```python
folder_url = "https://drive.google.com/drive/folders/1Ay5r96VRwcCl_sKwmb7-T81NlpgW8eRt?usp=drive_link"
folder_id = folder_url.split("/folders/")[1].split("?")[0]
folder_id
```
